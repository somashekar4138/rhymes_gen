# rhymes - sing a nursery rhyme on a free GPU

**Before you run anything: set the runtime to a T4 GPU.**

`Runtime` -> `Change runtime type` -> Hardware accelerator: **T4 GPU** -> Save.

Then `Runtime` -> `Run all`. Nothing below needs editing.

The first render downloads **22.4 GB** of model checkpoints
(15.75 GB of 3B weights, 6.64 GB of codec). That takes a while and is not a hang.
A second render in this same session reuses them and starts generating straight away.

## 3. Write your rhyme

## 5. Sing it

```python
!rhymes render sample_rhyme.txt -o sample_rhyme.mp3 --style nursery
```
